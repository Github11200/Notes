```python
import torch
from torch import nn
import matplotlib.pyplot as plt
```

```python
weight = 0.3
bias = 0.2

start = 0
step = 0.02
end = 1
X = torch.arange(start=start, step=step, end=end).unsqueeze(dim=1)
y = X * weight + bias

X, y
```

```python
train_split = int(0.8 * len(X))

X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]
```

```python
def plot_predictions(train_data=X_train,
                     train_label=y_train,
                     test_data=X_test,
                     test_label=y_test,
                     predictions=None):
    
    plt.scatter(train_data, train_label, c="g", s=4, label="Training data")

    plt.scatter(test_data, test_label, c="r", s=4, label="Test data")

    if predictions is not None:
        plt.scatter(test_data, predictions, c="b", s=4, label="Predictions")
    
    plt.legend(prop={ "size": 14 })
```

```python
plot_predictions()
```

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(1,
                                               requires_grad=True,
                                               dtype=float))

        self.bias = nn.Parameter(torch.randn(1,
                                               requires_grad=True,
                                               dtype=float))
    
    def forward(self, x):
        return x * self.weight + self.bias
```

```python
model_0 = LinearRegressionModel()

model_0.state_dict()
```
